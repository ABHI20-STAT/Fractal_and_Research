# Theoretical and Computational Analysis: Dynamic Polynomial Morphologies

## 1. Mathematical Scaling and Boundary Transitions

The morphological evolution of the degree-$d$ Multibrot set $\mathcal{M}_d$ is evaluated by tracking the critical point $z_0 = 0$ under the parameterized iteration sequence:

$$P_c(z) = z^d + c, \quad c \in \mathbb{C}$$

### Boundary Condensation Mechanics
As the polynomial degree map $d \in \mathbb{R}$ continuously scales upward, the topological features undergo an extreme phase transition:

* **Symmetry Multiplication ($C_{d-1}$)**: The fundamental polynomial framework generates exactly $d-1$ symmetrical critical vectors distributed evenly along the complex circle. This establishes an exact rotational cyclic group structure $C_{d-1}$ across the stable parameter boundary.
* **Asymptotic Disk Convergence**: For any exterior matrix coordinate where $|z| > 1$, the continuous power scaling forces $z_n^d$ to diverge to infinity at an exponential rate. Conversely, if $|z| < 1$, the value collapses directly to the origin. This mathematical threshold squashes the boundary layer tightly against the unit circle:

$$\lim_{d \to \infty} \partial\mathcal{M}_d = \mathbb{S}^1 = \{ c \in \mathbb{C} : |c| = 1 \}$$

### Continuous Potential Field Regularisation
To capture this dynamic thinning without visual stepping artifacts, the discrete step counters are converted into a continuous potential field $\Phi(c)$ using an analytical normalization scale:

$$\Phi(c) = \lim_{n \to \infty} \frac{\ln |z_n|}{d^n} \approx \frac{\ln |z_k|}{d^k}$$

where $k$ represents the exact index tracking vector where the critical path crosses the dynamic escape bailout circle ($|z_k| > R_d$).

# High-Performance Structural Animation Engine

## 1. Dynamic Frame Buffer Generation

To capture the transition as the polynomial degree $d$ grows without layout flickering, we construct a sequential frame rendering engine using `matplotlib.animation.FuncAnimation`.

### Microarchitectural Optimizations for Animation
* **Pre-allocated Plot Handles**: Instead of calling `plt.subplots()` or rebuilding layout cells inside the loop—which quickly exhausts system RAM and triggers performance bottlenecks—the animation canvas is initialized exactly once.
* **Direct Data Layer Mutation**: The animation callback targets the raster image buffer using `.set_data()`. It dynamically applies the analytical continuous potential equation step-by-step, shifting only the numerical array values under the hood.
* **Dynamic Title and Metric Updates**: Text labels update dynamically on every tick to show the current polynomial degree $d$ and its corresponding rotation symmetry group $C_{d-1}$ in real time.

## 2. Animation Script

Copy and paste this script directly into a standard code cell in your Google Colab notebook. It runs the entire vectorised computation loop, structures the frames, and encodes them into a crisp, standalone HTML5 video component that plays right inside your browser window.


```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import PowerNorm
from matplotlib.animation import FuncAnimation
from IPython.display import HTML, display
from pathlib import Path
import os

# --- ANIMATION CORE PROFILE CONFIGURATION ---
# Define a smooth fractional sequence of polynomial degrees to capture transition details
ANIMATION_DEGREES = np.linspace(2.0, 15.0, num=80)
GRID_RESOLUTION = 500  # Optimized for fast file compression
MAX_ITERS = 150

# Setup local folder architecture
OUTPUT_DIR = Path("/content/dissertation_outputs/animation_files")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

def compute_frame_potential(d, resolution=500, max_iter=150):
    """
    Computes an optimized smooth potential matrix for a fractional degree d.
    Fits the spatial grid dynamically as the shapes shrink toward the unit disk.
    """
    span = 1.45 if d <= 4.0 else 1.25

    x = np.linspace(-span, span, resolution)
    y = np.linspace(-span, span, resolution)
    X, Y = np.meshgrid(x, y)
    C = X + 1j * Y

    Z = np.zeros_like(C, dtype=np.complex128)
    smooth_potential = np.zeros(C.shape, dtype=np.float64)
    active = np.ones(C.shape, dtype=bool)

    bailout = max(3.0, (2.0**(1.0/(max(1.001, d)-1))) + 5.0)

    with np.errstate(invalid="ignore", over="ignore", divide="ignore"):
        for iteration in range(1, max_iter + 1):
            Z[active] = Z[active]**d + C[active]
            modulus = np.abs(Z)

            escaped = active & (~np.isfinite(modulus) | (modulus > bailout))
            if np.any(escaped):
                final_mod = np.where(np.isfinite(modulus[escaped]), modulus[escaped], bailout)
                smooth_potential[escaped] = iteration - (np.log(np.log(final_mod + 1e-10)) / np.log(max(1.001, d)))
                active[escaped] = False

            if not np.any(active):
                break

    return x, y, smooth_potential, active

# --- STEP 1: INITIALISE THE BASE CANVAS FRAME ---
print("Initializing base layout configurations...")
plt.ioff()  # Turn off immediate plot generation to keep the cell output clean

d_start = ANIMATION_DEGREES[0]
x_init, y_init, pot_init, unresolved_init = compute_frame_potential(d_start, GRID_RESOLUTION, MAX_ITERS)
extent_init = [x_init.min(), x_init.max(), y_init.min(), y_init.max()]

fig, ax = plt.subplots(figsize=(7, 6.5), constrained_layout=True)

display_matrix = np.ma.masked_where(unresolved_init, pot_init)
img_handle = ax.imshow(
    display_matrix,
    extent=extent_init,
    origin="lower",
    cmap="magma",
    norm=PowerNorm(gamma=0.45),
    interpolation="bilinear"
)

# Set base layout styling configurations
ax.set_xlabel(r"$\operatorname{Re}(c)$", fontsize=10)
ax.set_ylabel(r"$\operatorname{Im}(c)$", fontsize=10)
ax.set_aspect("equal", adjustable="box")
ax.patch.set_facecolor("#020508")  # Clean dark fill inside the bounded core regions
ax.grid(False)

# FIXED: Appended raw string marker r"" to completely isolate internal backslash symbols
cbar = fig.colorbar(img_handle, ax=ax, orientation="vertical", shrink=0.8, pad=0.03, aspect=30)
cbar.set_label(r"Analytical Potential Level $\Phi(c)$", fontsize=10, labelpad=8)

# --- STEP 2: DEFINE THE BUFFER SWAP FRAME MUTATOR ---
def update_animation_frame(frame_idx):
    """
    Directly updates the raw pixel memory buffers inside the image container
    to keep transitions clean and smooth.
    """
    d_current = ANIMATION_DEGREES[frame_idx]
    x_curr, y_curr, pot_curr, unresolved_curr = compute_frame_potential(d_current, GRID_RESOLUTION, MAX_ITERS)
    extent_curr = [x_curr.min(), x_curr.max(), y_curr.min(), y_curr.max()]

    next_matrix = np.ma.masked_where(unresolved_curr, pot_curr)

    img_handle.set_data(next_matrix)
    img_handle.set_extent(extent_curr)
    img_handle.set_clim(vmin=next_matrix.min(), vmax=next_matrix.max())

    ax.set_xlim(x_curr.min(), x_curr.max())
    ax.set_ylim(y_curr.min(), y_curr.max())

    # FIXED: Added raw string tags to the title wrapper string
    ax.set_title(
        f"Morphology Transition Profile\n"
        rf"Polynomial Degree $d = {d_current:.2f}$ | Symmetry Group: $C_{{{d_current-1:.2f}}}$",
        fontsize=11,
        pad=10
    )

    return [img_handle]

# --- STEP 3: ASSEMBLE AND RUN MULTI-FORMAT EXPORTER ENGINE ---
print(f"Compiling structural sequence framework ({len(ANIMATION_DEGREES)} individual frames)...")

multibrot_anim = FuncAnimation(
    fig,
    update_animation_frame,
    frames=len(ANIMATION_DEGREES),
    interval=60,
    blit=True
)

mp4_path = OUTPUT_DIR / "multibrot_evolution.mp4"
gif_path = OUTPUT_DIR / "multibrot_evolution.gif"

# 1. Attempt High-Fidelity MP4 Export via FFmpeg
print("Encoding MP4 video tracking file...")
try:
    multibrot_anim.save(str(mp4_path), writer='ffmpeg', fps=15, dpi=150)
    print(f"[SUCCESS] Native MP4 video export saved completely to:\n -> {mp4_path}")
except Exception as e:
    print(f"[WARNING] FFmpeg encoder unavailable. Skipping MP4 generation. Error: {e}")

# 2. Attempt Looping GIF Export via Pillow/ImageMagick
print("Encoding GIF frame animation loop...")
try:
    multibrot_anim.save(str(gif_path), writer='pillow', fps=15, dpi=150)
    print(f"[SUCCESS] Looping GIF graphic export saved completely to:\n -> {gif_path}")
except Exception as e:
    print(f"[WARNING] Pillow encoder ran into a processing error. Skipping GIF generation. Error: {e}")

# Display the interactive player window directly inside the workspace layout cells
print("\nLoading media engine inside your notebook output window...")
display(HTML(multibrot_anim.to_jshtml()))
```
